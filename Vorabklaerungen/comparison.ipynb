```python
#-------------------------------------------------------
#-- Titel       : Vergleich der Ergebnisse der FIR-Filter Implementierung in Python und VHDL
#-------------------------------------------------------
#-- Projekt     : Vorabklärung TB
#-- Filename    : plot_results.ipynb
#-- Autor       : Stefan Foltrauer
#-- Letze Änderungen : 17.10.2026
#-------------------------------------------------------
#-- Beschreibung: 
#-- Plotten der Ergebnisse der FIR-Filter Implementierung in Python und VHDL im Zeitbereich. Die Ergebnisse werden mit dem Originalsignal verglichen.
#-------------------------------------------------------
import numpy as np
import matplotlib.pyplot as plt

fs = 200

input_signal = np.loadtxt("input_signal.txt")
python_output = np.loadtxt("python_output.txt")
vhdl_output = np.loadtxt("vhdl_output.txt")


t = np.arange(len(input_signal)) / fs

plt.plot(t,input_signal,label="Original signal")
plt.plot(t,python_output,label="Python FIR")
plt.plot(t,vhdl_output,label="VHDL FIR")

plt.xlabel("Time /s")
plt.ylabel("Amplitude")

plt.legend()
plt.grid()
plt.show()
```
