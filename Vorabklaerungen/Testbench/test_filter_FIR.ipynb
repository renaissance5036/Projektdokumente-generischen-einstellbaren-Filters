```python
#-------------------------------------------------------
#-- Titel       : Pythonvergleichsfilter
#-------------------------------------------------------
#-- Projekt     : Vorabklärung TB
#-- Filename    : pythonvergleichsfilter.ipynb
#-- Autor       : Stefan Foltrauer
#-- Letze Änderungen : 17.10.2026
#-------------------------------------------------------
#-- Beschreibung: 
#-- Erstellt ein Vergleichsfilter in Python, das die gleiche Funktionalität wie der in VHDL implementierte Filter bietet.
#-------------------------------------------------------
import numpy as np

# FIR filter coefficients
Num = np.array([0.25,0.50,0.25])

# Input signal
input_signal = np.loadtxt("input_signal.txt", dtype=int)

# Output signal
output_signal = np.zeros(len(input_signal))

# Previous input values
x1 = 0
x2 = 0

for i in range(len(input_signal)):
    x0 = input_signal[i]

    output_signal[i] = (Num[0] * x0 + Num[1] * x1 + Num[2] * x2) #The filter as a impuls response
    x2 = x1
    x1 = x0
#create a new file with the outputsignal
np.savetxt("python_output.txt", output_signal, fmt="%d")
print("Saved to:", "python_output.txt")
```
